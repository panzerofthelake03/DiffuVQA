```python
# Clone the GitHub Repository
!git clone https://github.com/panzerofthelake03/DiffuVQA.git

# Change Directory to Project Folder
import os

project_path = '/content/DiffuVQA'
image_path = '/content/DiffuVQA/datasets/slake'
checkpoints_path = '/content/DiffuVQA/checkpoints'

image_set_url = "https://huggingface.co/datasets/BoKelvin/SLAKE/blob/main/imgs.zip"

os.chdir(project_path)
print("Current Directory:", os.getcwd())
```

```python
# Download and extract image dataset
import os
import zipfile
import requests

# Define paths
image_set_url = "https://huggingface.co/datasets/BoKelvin/SLAKE/resolve/main/imgs.zip"
image_zip_path = os.path.join(image_path, "imgs.zip")

# Create image directory if it doesn't exist
os.makedirs(image_path, exist_ok=True)

# Download the zip file
print("Downloading image dataset...")
response = requests.get(image_set_url, stream=True)
with open(image_zip_path, "wb") as f:
    for chunk in response.iter_content(chunk_size=1024):
        if chunk:
            f.write(chunk)

print("Download complete. Extracting files...")

# Extract the zip file
with zipfile.ZipFile(image_zip_path, "r") as zip_ref:
    zip_ref.extractall(image_path)

print("Extraction complete.")

# Clean up zip file
os.remove(image_zip_path)
print("Zip file removed.")
```
